### **Task 3: Trojan Analysis & Comparison**
Analyze and compare generated Trojans across different vulnerability types and designs.

```python
#@title Task 3: Trojan Analysis & Comparison
import glob
import re
import csv
from collections import defaultdict

def analyze_trojan_characteristics(output_dir="./batch_trojan_designs"):
    """Comprehensive analysis of generated Trojans"""

    print("🦠 TASK 3: Trojan Analysis & Comparison")
    print("=" * 50)

    if not os.path.exists(output_dir):
        print(f"❌ Output directory {output_dir} not found")
        return

    verilog_files = glob.glob(f"{output_dir}/**/*.v", recursive=True)
    taxonomy_files = glob.glob(f"{output_dir}/**/*_taxonomy.txt", recursive=True)

    print(f"📊 Found {len(verilog_files)} Trojaned designs")
    print(f"📋 Found {len(taxonomy_files)} taxonomy files")

    if not verilog_files:
        print("❌ No Trojaned designs found")
        return

    trojan_stats = defaultdict(list)
    vulnerability_analysis = defaultdict(lambda: defaultdict(list))
    all_rows = []

    for vfile in verilog_files:
        basename = os.path.basename(vfile)
        parts = basename.replace('.v', '').split('_')

        design_name = parts[0] if len(parts) > 0 else 'Unknown'
        vuln_type = next((p for p in parts if p.startswith('H')), 'Unknown')
        model_name = parts[parts.index(vuln_type) + 1] if vuln_type in parts and parts.index(vuln_type) + 1 < len(parts) else 'Unknown'

        with open(vfile, 'r') as f:
            content = f.read()

        lines = len(content.splitlines())
        always_blocks = len(re.findall(r'always\s*@', content))
        if_statements = len(re.findall(r'\bif\s*\(', content))
        registers = len(re.findall(r'\breg\s+', content))
        hardcoded_vals = len(re.findall(r"\d+'[bh][0-9a-fA-F_xXzZ]+", content))
        counters = len(re.findall(r'<=\s*\w+\s*\+\s*1', content))
        comparisons = len(re.findall(r'==\s*\d+', content))

        # Read trigger/payload from corresponding taxonomy file
        tax_path = vfile.replace('.v', '_taxonomy.txt')
        trigger, payload = '', ''
        if os.path.exists(tax_path):
            with open(tax_path, 'r') as f:
                tax = f.read()
            tm = re.search(r'Trigger:\s*(.+?)(?:\n\n|Payload:)', tax, re.DOTALL)
            pm = re.search(r'Payload:\s*(.+?)(?:\n\n|Taxonomy:)', tax, re.DOTALL)
            trigger = tm.group(1).strip().replace('\n', ' ') if tm else ''
            payload = pm.group(1).strip().replace('\n', ' ') if pm else ''

        stats = {
            'file': basename, 'design': design_name, 'vulnerability': vuln_type,
            'model': model_name, 'lines': lines, 'always_blocks': always_blocks,
            'if_statements': if_statements, 'registers': registers,
            'hardcoded_values': hardcoded_vals, 'counters': counters,
            'comparisons': comparisons, 'trigger': trigger, 'payload': payload
        }
        trojan_stats[vuln_type].append(stats)
        vulnerability_analysis[vuln_type][design_name].append(stats)
        all_rows.append(stats)

    # Print summary table
    print("\n" + "=" * 60)
    print("📊 TROJAN CHARACTERISTICS BY VULNERABILITY TYPE")
    print("=" * 60)

    for vuln_type, stats_list in trojan_stats.items():
        print(f"\n🦠 {vuln_type} ({len(stats_list)} samples):")
        avg = lambda key: sum(s[key] for s in stats_list) / len(stats_list)
        print(f"   Lines: {avg('lines'):.1f}  |  if-stmts: {avg('if_statements'):.1f}  |  regs: {avg('registers'):.1f}  |  counters: {avg('counters'):.1f}")
        if stats_list[0]['trigger']:
            print(f"   Sample trigger: {stats_list[0]['trigger'][:120]}")
        if stats_list[0]['payload']:
            print(f"   Sample payload: {stats_list[0]['payload'][:120]}")

    # Export summary.csv
    csv_path = "summary.csv"
    fieldnames = ['file', 'design', 'vulnerability', 'model', 'lines',
                  'always_blocks', 'if_statements', 'registers',
                  'hardcoded_values', 'counters', 'comparisons', 'trigger', 'payload']
    with open(csv_path, 'w', newline='') as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_rows)

    print(f"\n✅ Summary exported to {csv_path} ({len(all_rows)} rows)")

analyze_trojan_characteristics()
```

### **Task 4: Trojan Validation & Testing**
Validate that generated Trojans are functionally correct and test their trigger conditions.

```python
#@title Task 4: Trojan Validation & Testing
import subprocess
import tempfile
import glob
import os
import re

def validate_trojaned_designs(output_dir="./batch_trojan_designs", log_path="validation_log.txt"):
    """Validate syntax and basic functionality of generated Trojaned designs"""

    print("🦠 TASK 4: Trojan Validation & Testing")
    print("=" * 50)

    if not os.path.exists(output_dir):
        print(f"❌ Output directory {output_dir} not found")
        return

    verilog_files = sorted(glob.glob(f"{output_dir}/**/*.v", recursive=True))

    if not verilog_files:
        print("❌ No Trojaned designs found for validation")
        return

    print(f"🔍 Validating {len(verilog_files)} Trojaned designs...")

    validation_results = []
    log_lines = [f"Validation run — {len(verilog_files)} files\n{'='*60}"]

    for vfile in verilog_files:
        basename = os.path.basename(vfile)
        with open(vfile, 'r') as f:
            content = f.read()

        result = {
            'file': basename,
            'syntax_valid': None,
            'has_module': False,
            'has_trojans': False,
            'line_count': len(content.splitlines()),
            'issues': []
        }

        if 'module' in content and 'endmodule' in content:
            result['has_module'] = True
        else:
            result['issues'].append("Missing module/endmodule")

        if 'trojan' in content.lower():
            result['has_trojans'] = True

        suspicious_patterns = [
            r"==\s*\d+'[bh][0-9a-fA-F]+",
            r'<=\s*\w+\s*\+\s*1',
            r'if\s*\([^)]*==\s*\d+',
        ]
        pattern_count = sum(len(re.findall(p, content)) for p in suspicious_patterns)
        if pattern_count > 0:
            result['has_trojans'] = True

        try:
            with tempfile.NamedTemporaryFile(mode='w', suffix='.v', delete=False) as tmp:
                tmp.write(content)
                tmp_name = tmp.name
            rc = subprocess.call(['iverilog', '-t', 'null', tmp_name],
                                 stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
            result['syntax_valid'] = (rc == 0)
            if rc != 0:
                result['issues'].append("Syntax errors detected")
            os.unlink(tmp_name)
        except FileNotFoundError:
            result['syntax_valid'] = None  # iverilog not installed

        validation_results.append(result)

        status = "PASS" if result['has_module'] else "FAIL"
        iverilog_str = "PASS" if result['syntax_valid'] is True else ("FAIL" if result['syntax_valid'] is False else "SKIP")
        log_lines.append(
            f"{basename}: module={status}  iverilog={iverilog_str}  "
            f"trojan_markers={'YES' if result['has_trojans'] else 'NO'}  "
            f"lines={result['line_count']}  issues={result['issues'] or 'none'}"
        )

    # Print summary
    total = len(validation_results)
    valid_mod = sum(1 for r in validation_results if r['has_module'])
    valid_syn = sum(1 for r in validation_results if r['syntax_valid'] is True)
    checked_syn = sum(1 for r in validation_results if r['syntax_valid'] is not None)
    with_trojan = sum(1 for r in validation_results if r['has_trojans'])

    summary = [
        f"\n{'='*60}",
        f"Files validated:        {total}",
        f"Valid module structure: {valid_mod}/{total}",
        f"iverilog syntax pass:   {valid_syn}/{checked_syn} (skipped {total-checked_syn})",
        f"Trojan markers found:   {with_trojan}/{total}",
    ]
    for line in summary:
        print(line)
    log_lines.extend(summary)

    # Write log file
    with open(log_path, 'w') as f:
        f.write('\n'.join(log_lines) + '\n')
    print(f"\n✅ Validation log saved to {log_path}")

validate_trojaned_designs()
```
